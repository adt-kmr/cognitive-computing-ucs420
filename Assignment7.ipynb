```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
roll_no = 12345
np.random.seed(roll_no)
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

categories = ['Electronics', 'Clothing', 'Home & Kitchen', 'Sports']

sales_data = np.random.randint(1000, 5001, size=(12, 4))

df = pd.DataFrame(sales_data, columns=categories, index=months)

df
```

```python
df.head()
```

```python
df.describe()
```

```python
total_sales_category = df[categories].sum()

print("Total Sales per Category:")
print(total_sales_category)
```

```python
total_sales_month = df[categories].sum(axis=1)

print("Total Sales per Month:")
print(total_sales_month)
```

```python
sales_growth = df[categories].pct_change() * 100

average_sales_growth = sales_growth.mean()

print("Average Sales Growth for Each Category:")
print(average_sales_growth)
```

```python
df['Total Sales'] = df[categories].sum(axis=1)

df['Growth Rate'] = df['Total Sales'].pct_change() * 100

df
```

```python
if roll_no % 2 == 0:
    df['Electronics'] = df['Electronics'] * 0.90
else:
    df['Clothing'] = df['Clothing'] * 0.85

df
```

```python
plt.figure(figsize=(12, 6))

for category in categories:
    plt.plot(df.index, df[category], marker='o', label=category)

plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales Trends by Product Category')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[categories])

plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.title('Sales Distribution by Product Category')
plt.show()
```

```python
import numpy as np

array = np.array([
    [1, -2, 3],
    [-4, 5, -6]
])

array
```

```python
absolute_value = np.abs(array)

print(absolute_value)
```

```python
print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
```

```python
print("25th Percentile:", np.percentile(array, 25, axis=0))
print("50th Percentile:", np.percentile(array, 50, axis=0))
print("75th Percentile:", np.percentile(array, 75, axis=0))
```

```python
print("25th Percentile:", np.percentile(array, 25, axis=1))
print("50th Percentile:", np.percentile(array, 50, axis=1))
print("75th Percentile:", np.percentile(array, 75, axis=1))
```

```python
print("Mean:", np.mean(array))
print("Median:", np.median(array))
print("Standard Deviation:", np.std(array))
```

```python
print("Mean:", np.mean(array, axis=0))
print("Median:", np.median(array, axis=0))
print("Standard Deviation:", np.std(array, axis=0))
```

```python
print("Mean:", np.mean(array, axis=1))
print("Median:", np.median(array, axis=1))
print("Standard Deviation:", np.std(array, axis=1))
```

```python
my_list = [10, 20, 30, 40, 50]

print("Original List:", my_list)

pos1 = 1
pos2 = 3

temp = my_list[pos1]
my_list[pos1] = my_list[pos2]
my_list[pos2] = temp

print("List After Swapping:", my_list)
```

```python
my_set = {10, 20, 30, 40, 50}

print("Original Set:", my_set)

my_list = list(my_set)

pos1 = 1
pos2 = 3

my_list[pos1], my_list[pos2] = my_list[pos2], my_list[pos1]

my_set = set(my_list)

print("Set After Swapping:", my_set)
```
