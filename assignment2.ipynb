```python
# Q1: Create a list using the individual digits of the roll number

roll_number = 1024170314

# Convert roll number into individual digits
digits = [int(digit) for digit in str(roll_number)]

# Multiply each digit by 10
L = [digit * 10 for digit in digits]

print("Roll number:", roll_number)
print("Individual digits:", digits)
print("Initial list L:", L)
```

```python
# Add one number using append()
L.append(50)

# append() adds the new element at the end of the list.
print("After append(50):", L)

# Add another number using insert() at a specific position
L.insert(2, 25)

# insert(2, 25) adds 25 at index 2 and shifts the existing elements to the right.
print("After insert(2, 25):", L)
```

```python
# Remove an element using remove()
L.remove(10)

# remove(10) removes the first occurrence of the value 10 from the list.
print("After remove(10):", L)

# Remove an element using pop()
removed_element = L.pop(3)

# pop(3) removes and returns the element at index 3.
print("Element removed using pop():", removed_element)
print("After pop(3):", L)
```

```python
# Sort L in ascending order
L.sort()
print("Ascending order:", L)

# Sort L in descending order
L.sort(reverse=True)
print("Descending order:", L)
```

```python
# First three elements
print("First three elements:", L[:3])

# Last three elements
print("Last three elements:", L[-3:])
```

```python
# Calculate the average of all elements in L
average = sum(L) / len(L)

# Create a new list containing elements greater than the average
greater_than_average = [x for x in L if x > average]

print("Average of L:", average)
print("Elements greater than average:", greater_than_average)
```

```python
# Q2: Create a tuple using the first 8 values of the original list L.
# Note: L has been sorted in Q1, so these are the first 8 values
# of L at this stage.

scores = tuple(L[:8])

print("Scores tuple:", scores)
```

```python
# Find the highest score and its first index
highest_score = max(scores)
highest_index = scores.index(highest_score)

# Find the lowest score and its frequency
lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print("Highest score:", highest_score)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest_score)
print("Number of times lowest score appears:", lowest_count)
```

```python
# A tuple cannot be reversed in place because tuples are immutable.
# reversed() returns an iterator, which we convert into a list.

reversed_scores = list(reversed(scores))

print("Reversed tuple as a list:", reversed_scores)
```

```python
# Ask the user to enter a score
search_score = int(input("Enter a score to search for: "))

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score is not present in the tuple.")
```

```python
# Tuples are immutable, so their elements cannot be changed directly.

try:
    scores[0] = 100
except TypeError as error:
    print("Error:", error)
    print("A tuple cannot be modified after creation because it is immutable, unlike a list.")
```

```python
# Unpack the first score, second score, and all remaining scores
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```
